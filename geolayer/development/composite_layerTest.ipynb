### Add two or more vector layers

```python
importlib.reload(composite_layer)
importlib.reload(vector_layer)
from vois.geo import Map

vlayer1 = vector_layer.VectorLayer.file('/data/NUTS_RG_03M_2021_4326_0.shp')
vlayer1.symbologyAdd(rule="[CNTR_CODE] = 'IT'", symbol=[ [ ["PolygonSymbolizer", "fill", '#ff0000'] ] ])

vlayer2 = vector_layer.VectorLayer.file('/data/NUTS_RG_03M_2021_4326_0.shp')
vlayer2.symbologyAdd(rule="[CNTR_CODE] = 'FR'", symbol=[ [ ["PolygonSymbolizer", "fill", '#00ff00'] ] ])

vlayer3 = vector_layer.VectorLayer.file('/data/NUTS_RG_03M_2021_4326_0.shp')
vlayer3.symbologyAdd(rule="[CNTR_CODE] = 'DE'", symbol=[ [ ["PolygonSymbolizer", "fill", '#0000ff'] ] ])

clayer = composite_layer.CompositeLayer()
clayer.add(vlayer1)
clayer.add(vlayer2)
clayer.add(vlayer3)

#clayer.print()
#print(clayer.xml())
#print(clayer.tileUrl(file_format='png'))

m = Map.Map(zoom=5, center=[46.8, 11.56], basemapindex=0)
m.addLayer(clayer)
m.onclick = vlayer1.onclick
vlayer1.identify_fields = ['FID', 'CNTR_CODE']
m
```

### Composite of composites

```python
importlib.reload(composite_layer)
importlib.reload(raster_layer)
importlib.reload(vector_layer)
import plotly.express as px
from vois.geo import Map

# Masking of a raster over Italy
vlayer = vector_layer.VectorLayer.file('/data/NUTS_RG_03M_2021_4326_0.shp')
vlayer.symbologyAdd(rule="[CNTR_CODE] = 'IT'", symbol=[ [ ["PolygonSymbolizer", "fill", '#ff0000'] ] ])

rlayer = raster_layer.RasterLayer.single('/data/CAMX_LAMMA_2019_NO2_IT_TUSCANY-MAP_FILE-20240731-ee2be3-1167.tiff')
rlayer.colorlist(0.0, 10.0, px.colors.sequential.Viridis)

clayer1 = composite_layer.CompositeLayer()
clayer1.add(vlayer)
clayer1.add(rlayer, composition_operation='src-in')


# Mosaic of FR and DE
vlayer1 = vector_layer.VectorLayer.file('/data/NUTS_RG_03M_2021_4326_0.shp')
vlayer1.symbologyAdd(rule="[CNTR_CODE] = 'FR'", symbol=[ [ ["PolygonSymbolizer", "fill", '#00ff00'] ] ])

vlayer2 = vector_layer.VectorLayer.file('/data/NUTS_RG_03M_2021_4326_0.shp')
vlayer2.symbologyAdd(rule="[CNTR_CODE] = 'DE'", symbol=[ [ ["PolygonSymbolizer", "fill", '#0000ff'] ] ])

clayer2 = composite_layer.CompositeLayer()
clayer2.add(vlayer1)
clayer2.add(vlayer2)


# Composite of the two composites
clayer = composite_layer.CompositeLayer()
clayer.add(clayer1)
clayer.add(clayer2)

#clayer.print()
#print(clayer.xml())
#print(clayer.tileUrl(file_format='png'))

m = Map.Map(zoom=5, center=[46.8, 11.56], basemapindex=0)
m.addLayer(clayer)
m.onclick = rlayer.onclick
m
```
