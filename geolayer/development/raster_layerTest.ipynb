### SENTINEL-2 display (also with STAC integration)

##### Read the STAC info on a Sentinel-2 product

```python
import requests
import json
url = 'http://stac-api.cidsn.jrc.it:20008/services/stac-api/collections/EO.Copernicus.S2.L2A/items/S2A_MSIL2A_20230910T100601_N0509_R022_T32TQP_20230910T161500'
#url = 'http://stac-api.cidsn.jrc.it:20008/services/stac-api/collections/EO.Copernicus.S2.L2A/items/S2A_MSIL1C_20240901T100031_N0511_R122_T33TUJ_20240901T134329'
response = requests.get(url)
stacitem = json.loads(response.text)
#stacitem
```

##### Single band

```python
%run ./development/load_local.py -w
from geolayer import rasterlayer
from vois.geo import Map
import plotly.express as px

#ly = rasterlayer.sentinel2single(stacitem, scalemin=800, scalemax=2800)
ly = rasterlayer.sentinel2single('S2A_MSIL2A_20230910T100601_N0509_R022_T32TQP_20230910T161500', 'B04', scaling='near', scalemin=800, scalemax=2800, colorlist=px.colors.sequential.Viridis)

m = Map.Map(center=[43.696, 12.1179], zoom=9)
m.addLayer(ly)
m.onclick = ly.onclick
m
```
