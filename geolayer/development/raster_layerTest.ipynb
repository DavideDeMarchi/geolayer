##### Read the STAC info on a Sentinel-2 product

```python
import requests
import json
url = 'http://stac-api.cidsn.jrc.it:20008/services/stac-api/collections/EO.Copernicus.S2.L2A/items/S2A_MSIL2A_20230910T100601_N0509_R022_T32TQP_20230910T161500'
#url = 'http://stac-api.cidsn.jrc.it:20008/services/stac-api/collections/EO.Copernicus.S2.L2A/items/S2A_MSIL1C_20240901T100031_N0511_R122_T33TUJ_20240901T134329'
response = requests.get(url)
stacitem = json.loads(response.text)
#stacitem
```
