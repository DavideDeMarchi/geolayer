### identify: search the nearest feature to a point in lat/lon coordinates using tolerance (to be set greater than 0 for polylines and points!)

```python
# Shapefile (layer can be omitted!)
%run vectorAPI.py
res = identify('/data/NUTS_RG_03M_2021_4326_0.shp', lon=12, lat=43, tolerance=0, geom=False)
res
```

```python
# Shapefile points
%run vectorAPI.py
res = identify('/data/AirfldP.shp', lon=12, lat=43, tolerance=0.5, geom=False)
res
```

```python
# Geopackage
%run vectorAPI.py
res = identify('/data/IsleofMan.gpkg', layer_name='31311023', lon=-4.6575, lat=54.0767, tolerance=0.1, geom=False)
res
```

```python
# SQlite
%run vectorAPI.py
res = identify('/data/BA_effis.sqlite', layer_name='2016', lon=-8.0, lat=41.0, tolerance=0.1, geom=False)
res
```
